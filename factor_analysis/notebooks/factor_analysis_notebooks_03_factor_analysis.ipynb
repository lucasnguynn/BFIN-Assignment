## 2. Market Model (CAPM)
Estimate Alpha and Beta for the frozen portfolio against the market proxy.

```python
def run_capm(returns, mkt_rf):
    X = sm.add_constant(mkt_rf)
    model = sm.OLS(returns, X).fit()
    return model

# Example usage:
# capm_model = run_capm(portfolio_excess_returns, df_factors['Mkt-RF'])
# print(capm_model.summary())
```

## 3. Fama-French 3-Factor Model (Lecture 7)
Estimate exposures to Market, Size (SMB), and Value (HML) to determine Theta vs Eta implications.

```python
def run_ff3(returns, factors_df):
    X = sm.add_constant(factors_df[['Mkt-RF', 'SMB', 'HML']])
    model = sm.OLS(returns, X).fit()
    return model

# Example usage:
# ff3_model = run_ff3(portfolio_excess_returns, df_factors)
# print(ff3_model.summary())
```
